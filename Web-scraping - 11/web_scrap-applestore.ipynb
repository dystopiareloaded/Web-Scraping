```python
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
with open('apple_store.html','r', encoding='utf-8') as f:
    html = f.read()
```

```python
soup = BeautifulSoup(html, 'lxml')
```

```python
products_container = soup.find_all('div', class_= 'product')
```

```python
name, price, availability, rating, shipping = [] , [] , [] , [], []

for products in products_container:
    name.append(products.find('h3').text.strip())
    product_details = products.find_all('p')
    price.append(product_details[0].text.strip().split(':')[1])
    availability.append(product_details[1].text.strip().split(':')[1])
    rating.append(product_details[2].text.strip())
    shipping.append(product_details[3].text.strip().split(':')[1])
    
    
```

```python
df = pd.DataFrame({
    'Name': name,
    'Price': price,
    'Quantity Available': availability,
    'Ratings': rating,
    'Estimated Shipping': shipping
})
```
