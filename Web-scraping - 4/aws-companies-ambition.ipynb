```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import time
from tqdm import tqdm
from fake_useragent import UserAgent
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
url = 'https://www.ambitionbox.com/list-of-companies?page=1'
driver = webdriver.Chrome()
driver.get(url)
response = driver.page_source
```

```python
soup = BeautifulSoup(response, 'html.parser')
```

```python
company_list = soup.find_all('div', class_= 'companyCardWrapper')
```

```python
len(company_list)
```

```python
500*20
```

```python
name, company_type, location, ratings, reviews = [], [], [], [], []

for company in company_list:

    name_tag = company.h2.text.strip()
    name.append(name_tag)
    #print(name_tag)

    ratings_tag = company.find('div', class_= 'rating_text rating_text--md').text.strip()
    ratings.append(ratings_tag)
    #print(ratings_tag)

    reviews_tag = company.find('span', class_='companyCardWrapper__companyRatingCount').text.strip().split('(')[1][:-1]
    reviews.append(reviews_tag)
    #print(reviews_tag)

    x = company.find('span', class_='companyCardWrapper__interLinking').text.strip().split('| ')
    if x is not None:
        ctype_tag = x[0].strip()
        location_tag = x[1].strip()
    else:
        ctype_tag = 'No records'
        location_tag = 'Not available'

    company_type.append(ctype_tag)
    location.append(location_tag)
    
    #print(ctype_tag)
    #print(location_tag)

    #print('------------------------')
    
    
```
