```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
from fake_useragent import UserAgent
```

```python
ua = UserAgent()
headers = {"User-Agent": ua.random}

url = "https://www.imdb.com/chart/top"

response = requests.get(url, headers=headers)
```

```python
response.raise_for_status()
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
movies = soup.find_all('li', class_= 'ipc-metadata-list-summary-item')
#movies = soup.find('ul').find_all('div', class_= 'ipc-metadata-list-summary-item__c')
```

```python
ranks, movie_name, years, ratings = [] , [] , [] , []

for movie in movies:
    name = movie.find('div', class_= 'sc-2bbfc9e9-0 jUYPWY cli-children').find('a').get_text(strip=True).split(".")[1]
    rank = movie.find('div', class_= 'sc-2bbfc9e9-0 jUYPWY cli-children').find('a').get_text(strip=True).split(".")[0]
    year = movie.find('div', class_= 'sc-2bbfc9e9-6 cZkKPy cli-title-metadata').find('span').text
    #yr = movie.find('span', class_= 'sc-2bbfc9e9-7 jttFlJ cli-title-metadata-item').text
    rating = movie.find('span', class_= 'ipc-rating-star--rating').text

    ranks.append(rank)
    movie_name.append(name)
    years.append(year)
    ratings.append(rating)

    
    
```

```python
# Create DataFrame
df = pd.DataFrame({
    "Rank": ranks,
    "Title": movie_name,
    "Year": years,
    "Rating": ratings
})
```
