## Web-Scraping Website - "https://www.imdb.com/chart/top/?ref_=nv_mv_250"

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import time
from tqdm import tqdm
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
url = 'https://www.imdb.com/chart/top/?ref_=nv_mv_250'
#headers = {'user-agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/136.0.0.0 Safari/537.36'}
```

```python
driver = webdriver.Chrome()
driver.get(url)
button = driver.find_element(By.ID, 'list-view-option-detailed')
button.click()
```

```python
soup = BeautifulSoup(driver.page_source, 'html.parser')
```

```python
#response = requests.get(url, headers=headers)
```

```python
#soup = BeautifulSoup(response.text, 'lxml')
```

```python
movies_list = soup.find_all('li',class_= 'ipc-metadata-list-summary-item')
```

```python
len(movies_list)
```

```python
titles , images, years, durations, imdb_ratings, description = [], [], [], [], [], []

for movie in movies_list:

    title_tag = movie.h3.text.strip().split('. ')[1]
    titles.append(title_tag)

    img_tag = movie.img['src']
    images.append(img_tag)

    x = movie.find_all('span', {'class' : 'sc-4b408797-8 iurwGb dli-title-metadata-item'})
    if x:
        year_tag = x[0].text.strip()
        
    else:
        year_tag = "Year not found"
    years.append(year_tag)
    
    
    if x:
        duration_tag = x[1].text.strip()
        
    else:
    
        duration_tag = "Duration not found"
    durations.append(duration_tag)
    
    imdb_rating_tag = movie.find('span', class_='ipc-rating-star--rating').text.strip()
    imdb_ratings.append(imdb_rating_tag)

    desc_tag = movie.find('div', class_='ipc-html-content-inner-div').text.strip()
    description.append(desc_tag)

    
    
    
```

```python
titles
```

```python
df = pd.DataFrame({
    'Titles': titles,
    'Year' : years,
    'Duration': durations,
    'Images':images,
    'Description': description,
    'Ratings': imdb_ratings
})
```
