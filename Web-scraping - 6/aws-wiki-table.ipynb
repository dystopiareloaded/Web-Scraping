# Web-scraping of website - "https://en.wikipedia.org/wiki/List_of_cities_by_sunshine_duration"

```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://en.wikipedia.org/wiki/List_of_cities_by_sunshine_duration"
webpage = requests.get(url)
```

```python
soup = BeautifulSoup(webpage.text,'lxml')
```

```python
#print(soup.prettify())
```

```python
table = soup.find('table',{'class':'wikitable plainrowheaders sortable'})
```

```python
container = table.find_all('tr')
```

```python
country = []
city = []
year = []

for i in container[1:]:
    x = i.find_all('td')
    if len(x) >= 15:
        country.append(x[0].text.strip())
        city.append(x[1].text.strip())
        year.append(x[14].text.strip())
```

```python
year
```

```python
df = pd.DataFrame({
    'Country':country,
    'City': city,
    'Yearly Sunshine':year
    
})
```
