# Web-Scraping Website - "https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue"

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
#print(soup.prettify())
```

```python
print(len(soup.find('tbody')))
```

```python
print(len(soup.find('tbody').find_all('tr')))
```

```python
revenues = soup.find('tbody').find_all('tr')
```

```python
rank = []
name = []
headquarters = []

for revenue in revenues[1:]:
    cols = revenue.find_all('td')
    #for a_tag in cols:
        #print((a_tag).text)
    if len(cols) >=7:
        rank.append(cols[0].text.strip())
        name.append(cols[1].text.strip())
        headquarters.append(cols[-1].text.strip())
        
    
    
```

```python
data = {
    'Rank': rank,
    'Name': name,
    'Headquarters': headquarters
}

df_revenue = pd.DataFrame(data)
```
