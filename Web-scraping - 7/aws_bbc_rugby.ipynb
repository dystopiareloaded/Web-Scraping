# Web-Scraping Website - "https://www.bbc.com/sport/rugby-union/womens-six-nations/table"

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.bbc.com/sport/rugby-union/womens-six-nations/table'
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
print(soup.prettify())
```

```python
teams = soup.find('tbody').find_all('tr', class_= 'ssrcss-dhlz6k-TableRowBody e1icz100')
```

```python
team_names = []
points = []
played = []
won = []
lost = []

for team in teams:
    team_names.append(team.find('div', class_ = 'ssrcss-1m4x485-DesktopWidthOnly e1sk6csp2').get_text(strip=True))
    points.append(team.find('div', class_ = 'ssrcss-8k20kk-CellWrapper ef9ipf0').find('span').text)

    stats = team.find_all('span', class_= "visually-hidden ssrcss-1f39n02-VisuallyHidden e16en2lz0")  

    for div_tag in stats:
        print(div_tag)

    played.append(stats[1].text)
    won.append(stats[3].text)
    lost.append(stats[5].text)


    #for div_tag in stats:
     #   print(div_tag)
```

```python
points
```

```python
data = {
    'Team': team_names,
    'Played': played,
    'Won': won,
    'lost':lost,
    'Points Table': points
}
```

```python
df_rugby = pd.DataFrame(data)
```
