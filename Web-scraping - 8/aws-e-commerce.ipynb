# Web-scraping Websites - "https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"

```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"
response = requests.get(url)
response
```

```python
soup =  BeautifulSoup(response.text, "lxml")
#print(soup)
```

```python
boxes = soup.find_all('div', class_ = 'product-wrapper card-body')
```

```python
names = []
prices = []
info = []

for i in boxes:
    try:
        names.append(i.find('a', class_= 'title').text.strip())
    except: 
        names.append(np.nan)
    try:
        prices.append(i.find('span').text)
    except: 
        prices.append(np.nan) 
    try:
        info.append(i.find('p', class_= 'description card-text').text)
    except:
        info.append(np.nan)  
```

```python
info
```

```python
df = pd.DataFrame({
    'name': names,
    'price': prices,
    'description': info
})
```
