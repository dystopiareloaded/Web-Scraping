# Web-scraping website - "https://www.iplt20.com/auction/2022"

```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.iplt20.com/auction/2022"
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, 'lxml')
```

```python
table = soup.find("tbody").find_all('tr')
```

```python
print(len(table))
```

```python
team_names = []
funds = []
overseas_players = []
total_players = []

for i in table:
    team_names.append(i.find('h2', class_= 'ih-pt-cont').text)
    data = i.find_all('td')

    funds.append(data[2].text)
    overseas_players.append(data[3].text)
    total_players.append(data[4].text)

    #for td_tag in data:
     #   print(td_tag)
    
```

```python
overseas_players
```

```python
df = pd.DataFrame({
    'Team': team_names,
    'Funds': funds,
    'Overseas Players': overseas_players,
    'Total Players': total_players
})
```
