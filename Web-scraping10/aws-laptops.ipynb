## Web-scraping website - 'https://webscraper.io/test-sites/e-commerce/more/computers/laptops'

```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests
```

```python
with open("laptops.html", "r", encoding="utf-8") as f:
    html = f.read()
```

```python
soup = BeautifulSoup(html, 'lxml')
```

```python
products_divs = soup.find_all('div', class_='col-md-4 col-xl-4 col-lg-4')
```

```python
name , price, info = [] , [] , []

for product in products_divs:
    product_name = product.find('a', class_='title').text.strip()
    product_price = product.find('h4', class_='price float-end pull-right').text.strip()
    product_info = product.find('p', class_='card-text description').text.strip()
    
    name.append(product_name)
    price.append(product_price)
    info.append(product_info)
```

```python
# Create the DataFrame
df = pd.DataFrame({
    'Name': name,
    'Price': price,
    'Info': info
})
```

```python
df.to_csv('laptops.csv', index=False)
```
