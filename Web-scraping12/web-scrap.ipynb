```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

```python
with open("bookinghotels.html", "r", encoding="utf-8") as f:
    html = f.read()
```

```python
soup = BeautifulSoup(html, 'lxml')
```

```python
#print(soup.prettify())
```

```python
hotel_divs = soup.find_all('div', class_='c82435a4b8 a178069f51 a6ae3c2b40 a18aeea94d d794b7a0f7 f53e278e95 c6710787a4')
```

```python
hotels, prices, reviews = [], [], []

for hotel in hotel_divs:
    hotel_name = hotel.find('div', class_='f6431b446c a15b38c233').text.strip()
    price = hotel.find('span', class_='f6431b446c fbfd7c1165 e84eb96b1f').text.strip()

    x = hotel.find_all('div', class_='c624d7469d a0e60936ad a3214e5942 a0ff1335a1')
    review = x[0].text.strip() if x else 'No review'

    # Append to lists
    hotels.append(hotel_name)
    prices.append(price)
    reviews.append(review)
```

```python
# Create the DataFrame
df = pd.DataFrame({
    'Hotel': hotels,
    'Price': prices,
    'Review': reviews
})
```
